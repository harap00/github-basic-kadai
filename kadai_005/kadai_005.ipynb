```python
def calculate_trapezoid_area(jyouhen, kahen, takasa):
    """
    台形の面積を計算する関数
    公式: (上辺 + 下辺) * 高さ / 2
    """
    area = (jyouhen + kahen) * takasa / 2
    return area

# 定義した関数をテストします
jyouhen_val = 10
kahen_val = 20
takasa_val = 5

mensteki = calculate_trapezoid_area(jyouhen_val, kahen_val, takasa_val)
print(f"台形の面積: {mensteki} cm²")
```
