```python
class Human:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}さんは20歳以上（{self.age}歳）なので、大人です。")
        else:
            print(f"{self.name}さんは20歳未満（{self.age}歳）なので、大人ではありません。")
```

```python
# Humanクラスのインスタンスを複数生成してリストに追加
human_list = [
    Human("太郎", 25),
    Human("花子", 18),
    Human("次郎", 20),
    Human("三郎", 15)
]

# リストの要素数分だけcheck_adultメソッドを呼び出す
for human in human_list:
    human.check_adult()
```
