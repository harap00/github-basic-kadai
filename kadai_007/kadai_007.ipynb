```python
array = [
    "月曜日は晴れです",
    "火曜日は雨です",
    "水曜日は晴れです",
    "木曜日は晴れです",
    "金曜日は曇りです",
    "土曜日は曇りのち雨です",
    "日曜日は雷雨です"
]

# リストから水曜日の天気予報を表示
print(array[2])
```

```python
dictionary = {
    'mon': '晴れ',
    'tue': '雨',
    'wed': '晴れ',
    'thu': '晴れ',
    'fri': '曇り',
    'sat': '曇りのち雨',
    'sun': '雷雨'
}

# ディクショナリー構造を確認
for weekday in dictionary.keys():
    print(weekday, dictionary[weekday])
print()

# ディクショナリーから水曜日の天気を取り出す
print(dictionary['wed'])

```
