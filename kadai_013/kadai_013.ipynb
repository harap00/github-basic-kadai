```markdown
# 課題：消費税計算プログラム
商品の金額と消費税率を受け取り、消費税を加えた合計金額を計算して返す関数を作成します。
```

````python
def calculate_tax_inclusive_price(price, tax_rate):
    """
    商品の金額と消費税（%）から、消費税を加えた合計金額を計算する関数
    """
    # 消費税額を計算し、元の金額に加える（端数は切り捨てとします）
    total_price = int(price * (1 + tax_rate / 100))
    return total_price

# 引数の設定: 第1引数に商品金額「110」、第2引数に消費税「10」
item_price = 110
tax_percentage = 10

# 関数の実行と戻り値の受け取り
result = calculate_tax_inclusive_price(item_price, tax_percentage)

# 結果の表示
print(f"商品の金額: {item_price}円")
print(f"消費税率: {tax_percentage}%")
print(f"税込金額: {result}円")
````
